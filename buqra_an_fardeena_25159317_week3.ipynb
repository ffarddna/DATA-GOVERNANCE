```python
import pandas as pd 
import numpy as np 
# Load a sample dataset directly from a public URL 
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv' 
# Read the CSV file and store it in a DataFrame (a table) called 'df' 
df = pd.read_csv(url) 
print("Data loaded successfully!") 
```

```python
# 1. Look at the first 5 rows to see what the data looks like 
print("--- First 5 Rows ---") 
display(df.head()) 
# 2. Get a technical summary (column names, data types, and total rows) 
print("\n--- Data Information ---") 
df.info() 
# 3. Check for exact missing values (NaN) in each column 
print("\n--- Missing Values Count ---") 
print(df.isnull().sum())
```

```python
# OPTION A: Drop rows where 'Embarked' is missing 
# Since only 2 passengers are missing this data, it is safe to just delete those 2 rows. 
df_clean = df.dropna(subset=['Embarked']).copy() 
# OPTION B: Impute (fill) missing 'Age' values with the median age 
# Since many ages are missing, deleting them would ruin our dataset.  
# Instead, we find the middle (median) age and fill the blanks with it. 
median_age = df_clean['Age'].median() 
df_clean['Age'] = df_clean['Age'].fillna(median_age) 
# Let's check our missing values again to see if it worked! 
print("Missing values after cleaning:") 
print(df_clean[['Age', 'Embarked']].isnull().sum()) 
```

```python
# Drop the Cabin column from the cleaned dataset
df_clean = df_clean.drop(columns=['Cabin'])

# Display the remaining column names
print(df_clean.columns)
```
